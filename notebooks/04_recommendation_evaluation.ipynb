# 04. Recommendation Evaluation & Benchmarking
**Project**: Next-Best-Offer and Product Recommendation with Supply Constraints

Documents the out-of-time evaluation methodology, Precision@K / Recall@K / HitRate@K, and comparative visual benchmarks from actual recorded evaluation results.

## Evaluation Methodology
- **Temporal split**: Training window `< 2026-06-01`; evaluation window `>= 2026-06-01`.
- **Leakage prevention**: Features are computed strictly before the cutoff date.
- **Evaluation population**: Repeat customers with purchase history pre-cutoff and actual purchases in the evaluation window.


```
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
REPORTS_DIR = Path('../reports')

# Load the actually recorded evaluation results
eval_path = REPORTS_DIR / 'day3_evaluation_results.csv'
if eval_path.exists():
    eval_df = pd.read_csv(eval_path)
    print('=== Recorded Evaluation Results (from reports/day3_evaluation_results.csv) ===')
    print(eval_df.to_string(index=False))
else:
    print(f'[Not found] {eval_path}')
    print('Run: python scripts/evaluate_recommender.py')
```

### Comparative Visual Benchmarks: Day-2 Baseline vs. Day-3 Decision Pipeline

```
if eval_path.exists():
    fig, axes = plt.subplots(1, 3, figsize=(16, 5))

    # 1. Precision@K Comparison
    sns.barplot(data=eval_df, x='K', y='Precision@K', hue='Pipeline', palette='Set2', ax=axes[0])
    axes[0].set_title('Precision@K Comparison', fontsize=12, fontweight='bold')
    axes[0].set_ylabel('Precision@K')
    axes[0].set_ylim(0, 0.20)

    # 2. Recall@K Comparison
    sns.barplot(data=eval_df, x='K', y='Recall@K', hue='Pipeline', palette='Set2', ax=axes[1])
    axes[1].set_title('Recall@K Comparison', fontsize=12, fontweight='bold')
    axes[1].set_ylabel('Recall@K')
    axes[1].set_ylim(0, 0.25)

    # 3. HitRate@K Comparison
    sns.barplot(data=eval_df, x='K', y='HitRate@K', hue='Pipeline', palette='Set2', ax=axes[2])
    axes[2].set_title('HitRate@K Comparison', fontsize=12, fontweight='bold')
    axes[2].set_ylabel('HitRate@K')
    axes[2].set_ylim(0, 0.60)

    plt.tight_layout()
    plt.show()
```

### Limitations
- Evaluation is conducted on a 300-customer, 52-product synthetic ecosystem. Real-world scale will require Approximate Nearest Neighbor indexing.
- Cold-start customers fall back to popularity baseline.
- Association rules and collaborative filtering are not yet implemented (stubs exist in `backend/app/recommendation/`).
